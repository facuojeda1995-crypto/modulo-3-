```python
import pandas as pd
import os

# Creamos el archivo CSV de transacciones
df_trans_temp = pd.DataFrame({
    'id_producto': [101, 102, 103, 101, 104, 102],
    'cantidad': [2, 5, 1, 3, 10, 4],
    'fecha': ['2026-06-01', '2026-06-02', '2026-06-03', None, '2026-06-05', '2026-06-06']
})
df_trans_temp.to_csv("transacciones.csv", index=False)

# Creamos el archivo Excel de productos
df_prod_temp = pd.DataFrame({
    'id_producto': [101, 102, 103, 104],
    'nombre_producto': ['Laptop', 'Mouse', 'Teclado', 'Monitor'],
    'precio_unitario': [1200.0, 25.0, 45.0, 300.0]
})
df_prod_temp.to_excel("productos.xlsx", index=False)

print("¡Archivos 'transacciones.csv' y 'productos.xlsx' generados con éxito en Colab!")
```

```python
# 1. Adquisición de datos desde los archivos
df_transacciones = pd.read_csv("transacciones.csv")
df_productos = pd.read_excel("productos.xlsx")

# 2. Transformación y Limpieza
# Convertir fechas a datetime y eliminar nulos críticos
df_transacciones['fecha'] = pd.to_datetime(df_transacciones['fecha'])
df_transacciones = df_transacciones.dropna(subset=['id_producto', 'fecha'])

# Asegurar tipos de datos idénticos en la llave de unión
df_transacciones['id_producto'] = df_transacciones['id_producto'].astype(int)
df_productos['id_producto'] = df_productos['id_producto'].astype(int)

# Realizar el merge (unión) usando el id_producto
df_merged = pd.merge(df_transacciones, df_productos, on='id_producto', how='inner')
df_merged = df_merged.dropna(subset=['cantidad', 'precio_unitario'])

# 3. Normalización: Crear la columna calculada 'total_venta'
df_merged['total_venta'] = df_merged['cantidad'] * df_merged['precio_unitario']

print("--- DATASET FINAL NORMALIZADO ---")
display(df_merged)

# 4. Exportación a formato Parquet
archivo_salida = "ventas_procesadas.parquet"
df_merged.to_parquet(archivo_salida, index=False)
print(f"\n¡Archivo optimizado guardado exitosamente como '{archivo_salida}'!")
```
